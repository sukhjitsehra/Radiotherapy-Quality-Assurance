```python
class Field():
    def __init__(self, file_name):
        import re
        self.file = file_name

        pattern_new = re.compile("FieldID_\d{7}.json")

        pattern_pkl = re.compile("Field\d{7}.pkl")
        pattern_json = re.compile("Field\d{7}.json")

        if pattern_new.search(self.file) is not None and \
        self.file.endswith(pattern_new.search(self.file)[0]):

            self.field_info = self.__import_field_info(file_name)
            
            self.beamEnergy = self.field_info["beamEnergy"]
            self.beamMeterset = self.field_info['beamMeterset']
            self.bolusStatus = self.field_info['bolusStatus']
            self.description = self.field_info['description']
            self.fieldId = self.field_info['fieldId']
            self.hasBolus = self.field_info['hasBolus']
            self.isFff = self.field_info['isFff']
            self.isocentre = self.field_info['isocentre']
            self.machineName = self.field_info['machineName']
            self.machineType = self.field_info['machineType']

            self.mqType = self.field_info['mqType']
            self.name = self.field_info['name']
            self.type = self.field_info['type']
            self.__set_control_points()
        
        elif pattern_pkl.search(self.file) is not None and \
        self.file.endswith(pattern_pkl.search(self.file)[0]):
            import pickle
            #start = time.time()
            with open(self.file,"rb") as f:
                self = pickle.load(f)
               # print("runtime: ", time.time()-start)

        elif pattern_json.search(self.file) is not None and \
        self.file.endswith(pattern_json.search(self.file)[0]):
            import json, pandas as pd
            
            with open(self.file, "r") as f:
                dictionairy = json.load(f)
            self.file = dictionairy["file"]
            self.beamEnergy = dictionairy["beamEnergy"]
            self.beamMeterset = dictionairy["beamMeterset"]
            self.bolusStatus = dictionairy["bolusStatus"]
            self.description = dictionairy["description"]
            self.fieldId = dictionairy["fieldId"]
            self.hasBolus = dictionairy["hasBolus"]
            self.isFff = dictionairy["isFff"]
            self.isocentre = dictionairy["isocentre"]
            self.machineName = dictionairy["machineName"]
            self.machineType = dictionairy["machineType"]
            self.mqType = dictionairy["mqType"]
            self.name = dictionairy["name"]
            self.type = dictionairy["type"]
            self.leafSetA = pd.DataFrame(dictionairy["leafSetA"])
            self.leafSetB = pd.DataFrame(dictionairy["leafSetB"])
            self.collimatorAngle = dictionairy["collimatorAngle"]
            self.collimatorX1 = dictionairy["collimatorX1"]
            self.collimatorX2 = dictionairy["collimatorX2"]
            self.collimatorY1 = dictionairy["collimatorY1"]
            self.collimatorY2 = dictionairy["collimatorY2"]
            self.couchAngle = dictionairy["couchAngle"]
            self.cumulativeMetersetWeight = dictionairy["cumulativeMetersetWeight"]
            self.fieldX = dictionairy["fieldX"]
            self.fieldY = dictionairy["fieldY"]
            self.mlcLeaves = dictionairy["mlcLeaves"]
            self.photonBeamEnergy = dictionairy["photonBeamEnergy"]
            self.point = dictionairy["point"]
            self.gantryAngle = list(dictionairy["cumulativeMetersetWeight"].keys())

    
    def __import_field_info(self, file_name):
        """
        imports the field information from the file name

        ------------------------------------------------
        Parameters
            file_name (str) - the name of the file to import from

        Returns
            field_info (dict) - the information for the field

        """
        import json

        if not isinstance(file_name, str):
            print("Error __import_field: file_name must be a string")

        if not file_name.endswith(".json"):
            print("Error __import_field: make sure file is json format")

        with open(file_name, "r") as f:
            field_info = json.load(f)

        return field_info
    
    def __identical(self, list):
        """
        Checks if all the values in the list are the same
        
        """
        return len(set(list)) == 1

    def __change_var(self, list):
        """
        If all the items in the list are the same then the functions returns
        that item, otherwise it returns the list
        """
        if self.__identical(list):
            return list[0]
        return list
    
    def __set_control_points(self):

        import pandas as pd 

        mqControlPoints = self.field_info['mqControlPoints']
        collimatorAngle_list = []
        collimatorX1_list = []
        collimatorX2_list = []
        collimatorAngle_list = []
        collimatorX1_list =[]
        collimatorX2_list = []
        collimatorY1_list = []
        collimatorY2_list = []
        couchAngle_list =[]
        cumulativeMetersetWeight_list =[]
        fieldX_list =[]
        fieldY_list = []
        gantryAngle_list = []
        leafSetA_list =[]
        leafSetB_list = []
        mlcLeaves_list = []
        photonBeamEnergy_list = []
        point_list = []

        for i in range(len(mqControlPoints)):
            collimatorAngle = mqControlPoints[i]["collimatorAngle"]
            collimatorX1 = mqControlPoints[i]["collimatorX1"]
            collimatorX2 = mqControlPoints[i]['collimatorX2']
            collimatorY1 = mqControlPoints[i]['collimatorY1']
            collimatorY2 = mqControlPoints[i]['collimatorY2']
            couchAngle = mqControlPoints[i]['couchAngle']
            cumulativeMetersetWeight = mqControlPoints[i]['cumulativeMetersetWeight']
            fieldX = mqControlPoints[i]['fieldX']
            fieldY = mqControlPoints[i]['fieldY']
            gantryAngle = mqControlPoints[i]['gantryAngle']
            leafSetA = mqControlPoints[i]['leafSetA']
            leafSetB = mqControlPoints[i]['leafSetB']
            mlcLeaves = mqControlPoints[i]['mlcLeaves']
            photonBeamEnergy = mqControlPoints[i]['photonBeamEnergy']
            point = mqControlPoints[i]['point']

            collimatorAngle_list.append(collimatorAngle)
            collimatorX1_list.append(collimatorX1)
            collimatorX2_list.append(collimatorX2)
            collimatorY1_list.append(collimatorY1)
            collimatorY2_list.append(collimatorY2)
            couchAngle_list.append(couchAngle)
            cumulativeMetersetWeight_list.append(cumulativeMetersetWeight)
            fieldX_list.append(fieldX)
            fieldY_list.append(fieldY)
            gantryAngle_list.append(gantryAngle)
            leafSetA_list.append(leafSetA)
            leafSetB_list.append(leafSetB)
            mlcLeaves_list.append(mlcLeaves)
            photonBeamEnergy_list.append(photonBeamEnergy)
            point_list.append(point)

        
        self.collimatorAngle = self.__change_var(collimatorAngle_list)
        self.collimatorX1 = self.__change_var(collimatorX1_list)
        self.collimatorX2= self.__change_var(collimatorX2_list)
        self.collimatorY1= self.__change_var(collimatorY1_list)
        self.collimatorY2= self.__change_var(collimatorY2_list)
        self.couchAngle= self.__change_var(couchAngle_list)
        self.cumulativeMetersetWeight= self.__change_var(cumulativeMetersetWeight_list)
        self.fieldX= self.__change_var(fieldX_list)
        self.fieldY= self.__change_var(fieldY_list)
        self.gantryAngle= self.__change_var(gantryAngle_list)
        self.leafSetA= pd.DataFrame(leafSetA_list).T
        self.leafSetB = pd.DataFrame(leafSetB_list).T
        self.mlcLeaves= self.__change_var(mlcLeaves_list)
        self.photonBeamEnergy= self.__change_var(photonBeamEnergy_list)
        self.point= self.__change_var(point_list)

        return 
    
    def save_to_pickle(self):
        import pickle 
        file_name = "Field" + str(self.fieldId)
        with open(file_name + '.pkl', "wb") as file:
            pickle.dump(self, file)

    def __to_dict(self, values, keys):
        if not isinstance(keys, list):
            keys = range(len(values))
        if not isinstance(values, list):
            return values
        return dict(zip(keys, values))
    
    def save_to_json(self):
        import json
        file_name = "Field" + str(self.fieldId)

        dictionairy = {}

        dictionairy["file"] = self.file
        dictionairy["beamEnergy"] = self.beamEnergy
        dictionairy["beamMeterset"] = self.beamMeterset
        dictionairy["bolusStatus"] = self.bolusStatus
        dictionairy["description"] = self.description
        dictionairy["fieldId"] = self.fieldId
        dictionairy["hasBolus"] = self.hasBolus
        dictionairy["isFff"] = self.isFff
        dictionairy["isocentre"] = self.isocentre
        dictionairy["machineName"] = self.machineName
        dictionairy["machineType"] = self.machineType
        dictionairy["mqType"] = self.mqType
        dictionairy["name"] = self.name
        dictionairy["type"] = self.type

        dictionairy["leafSetA"] = leafSetA.to_dict()
        dictionairy["leafSetB"] = leafSetB.to_dict()

        dictionairy["collimatorAngle"] = self.__to_dict(self.collimatorAngle, self.gantryAngle)
        dictionairy["collimatorX1"] = self.__to_dict(self.collimatorX1, self.gantryAngle)
        dictionairy["collimatorX2"] = self.__to_dict(self.collimatorX2, self.gantryAngle)
        dictionairy["collimatorY1"] = self.__to_dict(self.collimatorY1, self.gantryAngle)
        dictionairy["collimatorY2"] = self.__to_dict(self.collimatorY2, self.gantryAngle)
        dictionairy["couchAngle"] = self.__to_dict(self.couchAngle, self.gantryAngle)
        dictionairy["cumulativeMetersetWeight"] = self.__to_dict(self.cumulativeMetersetWeight, self.gantryAngle)
        dictionairy["fieldX"] = self.__to_dict(self.fieldX, self.gantryAngle)
        dictionairy["fieldY"] = self.__to_dict(self.fieldY, self.gantryAngle)
        dictionairy["mlcLeaves"] = self.__to_dict(self.mlcLeaves, self.gantryAngle)
        dictionairy["photonBeamEnergy"] = self.__to_dict(self.photonBeamEnergy, self.gantryAngle)
        dictionairy["point"] = self.__to_dict(self.point, self.gantryAngle)

        with open(file_name + '.json', "w") as file:
            json.dump(dictionairy, file)
        
        return
```
