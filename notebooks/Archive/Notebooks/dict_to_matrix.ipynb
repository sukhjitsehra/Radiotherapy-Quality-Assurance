```python
# create a function for converting a dictionairy field to a matrix 

def dict_to_matrix(control_points):
    """
    converts the field information from dictionairy for to a matrix with
    gantry angles indexing the columns and leaf numbers indexing the rows
    """

    import pandas as pd
    import warnings
    warnings.simplefilter(action='ignore', category=FutureWarning) #added to suppress warnings

    matrixA = pd.DataFrame()
    matrixB = pd.DataFrame()

    for segment in control_points:
        matrixA[segment["gantryAngle"]] = segment["leafSetA"]
        matrixB[segment["gantryAngle"]] = segment["leafSetB"]

    # sort the columns so angle goes from 1 - 359
    matrixA = matrixA.reindex(sorted(matrixA.columns), axis=1)
    matrixB = matrixB.reindex(sorted(matrixB.columns), axis=1) 

    return matrixA, matrixB
```

```python
matrices = dict_to_matrix(field_info)
matrices.keys()
```

```python
# new calculation for calculating area by taking the sum over rows and columns

lengths = matrices["openinglength"]
gantry_areas = lengths.sum()*0.5
leaf_areas = lengths.T.sum()*0.5
average_gantry_areas = lengths.mean()*0.5
average_leaf_areas = lengths.T.mean()*0.5
```
