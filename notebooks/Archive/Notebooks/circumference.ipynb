```python

from field import Field
import os
```

```python
field = Field(os.path.dirname(os.getcwd()) +"\\EDA/testfield.json")
```

```python
rightA = field.matrixA.T.iloc[:,1:]
rightA.columns = field.matrixA.T.columns[:-1]
rightA.head()
```

```python
leftA = field.matrixA.T.iloc[:, :-1]
leftA.columns = field.matrixA.T.columns[1:]
leftA.head()
```

```python
matrixA = field.matrixA.T
left_diffA = abs(matrixA.iloc[:,1:] - leftA) 
left_diffA[field.active_leaf_start] = 0
left_diffA.head()
```

```python
right_diffA = abs(matrixA.iloc[:,:-1] - rightA)
right_diffA[field.active_leaf_end] = 0 
right_diffA.head()
```

```python
(left_diffA + right_diffA).head()
```

```python
circumferenceA  = 1/2*(left_diffA+ right_diffA) + 0.5
circumferenceA.head()
```

```python
rightB = field.matrixB.T.iloc[:,1:]
rightB.columns = field.matrixB.T.columns[:-1]

leftB = field.matrixB.T.iloc[:, :-1]
leftB.columns = field.matrixB.T.columns[1:]

matrixB = field.matrixB.T
left_diffB = abs(matrixB.iloc[:,1:] - leftB) 
left_diffB[field.active_leaf_start] = 0

right_diffB = abs(matrixB.iloc[:,:-1] - rightB)
right_diffB[field.active_leaf_end] = 0 

circumferenceB  = 1/2*(left_diffB+ right_diffB) + 0.5
circumferenceB.head()
```

```python
total_circumference = circumferenceA + circumferenceB
total_circumference.head()
```
